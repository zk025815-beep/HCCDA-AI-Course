```python
import pandas as pd
import os

# Correct the csv_file path based on the actual file found in the dataset
csv_file = os.path.join(path, "spotify-tracks-dataset-detailed.csv")

df = pd.read_csv(csv_file)
```

```python
def spotify_pipeline(df):

    # Make a copy
    df = df.copy()

    # 1. BASIC EXPLORATION
    print("Original Shape:", df.shape)

    print("\nMissing Values:")
    print(df.isnull().sum())

    print("\nDuplicate Rows:")
    print(df.duplicated().sum())

    print("\nStatistics:")
    print(df.describe())

    # 2. CLEANING
    df = df.drop_duplicates()

    df = df.dropna()

    # 3. CATEGORICAL ENCODING
    if "playlist_genre" in df.columns:
        df = pd.get_dummies(
            df,
            columns=["playlist_genre"],
            drop_first=True
        )

    # 4. CORRELATION
    correlation = df.corr(
        numeric_only=True
    )

    print("\nCorrelation Matrix:")
    print(correlation)

    # 5. FINAL INFORMATION
    print("\nFinal Shape:", df.shape)

    return df

cleand_data = spotify_pipeline(df)
```
