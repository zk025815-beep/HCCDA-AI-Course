```python
import cv2
import numpy as np
from google.colab import files
from google.colab.patches import cv2_imshow

# Upload an image from your computer
uploaded = files.upload()

# Get the uploaded image's filename
filename = next(iter(uploaded))

# Read the image using OpenCV
img = cv2.imread(filename)

# Display the original image
cv2_imshow(img)
```

Image dimensions and pixel values

```python
# Get image dimensions
height, width, channels = img.shape

print("Height:", height)
print("Width:", width)
print("Color channels:", channels)

# Get the pixel value at position (0, 0)
print("First pixel BGR value:", img[0, 0])
```

Color spaces (BGR/RGB)

```python
# OpenCV reads images in BGR format
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

print("Original BGR image:")
cv2_imshow(img)

print("Converted RGB image:")
cv2_imshow(rgb)
```

```python
# Convert the color image to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

cv2_imshow(gray)
```

```python
# Convert grayscale pixels into black or white
# Pixels above 127 become white; others become black
_, binary = cv2.threshold(
    gray, 127, 255, cv2.THRESH_BINARY
)

cv2_imshow(binary)
```

```python
# Detect edges and object boundaries
edges = cv2.Canny(gray, 100, 200)

cv2_imshow(edges)
```

```python
# Apply a blur to smooth the image
blurred = cv2.GaussianBlur(img, (5, 5), 0)

cv2_imshow(blurred)
```

```python
# Crop a section from the image
# Format: image[y_start:y_end, x_start:x_end]
cropped = img[0:200, 0:200]

cv2_imshow(cropped)
```

```python
# Resize the image to 300 pixels wide and 200 pixels high
resized = cv2.resize(img, (300, 200))

cv2_imshow(resized)
```

```python
# Save the processed images to files
cv2.imwrite("grayscale.jpg", gray)
cv2.imwrite("threshold.jpg", binary)
cv2.imwrite("edges.jpg", edges)
cv2.imwrite("blurred.jpg", blurred)
cv2.imwrite("cropped.jpg", cropped)
cv2.imwrite("resized.jpg", resized)

print("All processed images have been saved!")
```
