```python
# Load feedback data
import pandas as pd

feedback_data = {
    "feedback": [
        "The teacher explains concepts very clearly.",
        "The lectures are too fast and difficult to follow.",
        "Assignments are confusing and too difficult.",
        "The computer lab has slow internet.",
        "The teacher is helpful and answers our questions.",
        "We need more practical coding sessions.",
        "The classroom is clean and comfortable.",
        "The course workload is too high.",
        "The teaching methods are excellent.",
        "The lab computers are old and slow."
    ]
}

df = pd.DataFrame(feedback_data)

print(df.head())
print("Total feedback comments:", len(df))
```

```python
# Text preprocessing
import re

def clean_text(text):
    text = text.lower()
    text = re.sub(r"[^a-z\s]", "", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

df["clean_feedback"] = df["feedback"].apply(clean_text)

print(df[["feedback", "clean_feedback"]].head())
```

```python
positive_words = [
    "excellent", "clearly", "helpful",
    "comfortable", "clean", "good"
]

negative_words = [
    "difficult", "confusing", "slow",
    "fast", "high", "poor"
]

def analyze_sentiment(text):
    words = text.split()

    positive_count = sum(
        word in positive_words for word in words
    )

    negative_count = sum(
        word in negative_words for word in words
    )

    if positive_count > negative_count:
        return "Positive"
    elif negative_count > positive_count:
        return "Negative"
    else:
        return "Neutral"

df["sentiment"] = df["clean_feedback"].apply(analyze_sentiment)
```

```python
topic_keywords = {
    "Teaching": [
        "teacher", "teaching", "lectures", "concepts"
    ],
    "Assignments": [
        "assignments", "workload", "course"
    ],
    "Facilities": [
        "lab", "internet", "classroom", "computers"
    ],
    "Practical Learning": [
        "practical", "coding", "sessions"
    ]
}

def identify_topic(text):
    scores = {}

    for topic, keywords in topic_keywords.items():
        scores[topic] = sum(
            keyword in text for keyword in keywords
        )

    best_topic = max(scores, key=scores.get)

    if scores[best_topic] == 0:
        return "Other"

    return best_topic

df["topic"] = df["clean_feedback"].apply(identify_topic)

print(df[["feedback", "topic", "sentiment"]])
```

```python
negative_feedback = df[
    df["sentiment"] == "Negative"
]

concerns = negative_feedback["topic"].value_counts()

print("Most common negative-feedback topics:")
print(concerns)

print("\nAll topic counts:")
print(df["topic"].value_counts())
```

```python
print("SMART STUDENT FEEDBACK REPORT")
print("=" * 35)

print("Total comments:", len(df))

print("\nSentiment distribution:")
print(df["sentiment"].value_counts())

print("\nFeedback by topic:")
print(df["topic"].value_counts())

print("\nNegative comments by topic:")
print(
    df[df["sentiment"] == "Negative"]["topic"]
    .value_counts()
)

print("\nSample analyzed feedback:")
print(
    df[["feedback", "sentiment", "topic"]]
    .head(10)
    .to_string(index=False)
)
```

```python
df.to_csv("student_feedback_analysis.csv", index=False)

print("Analysis saved successfully.")
```
